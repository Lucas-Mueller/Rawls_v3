```python
import anthropic
from anthropic import AnthropicVertex

# Set your Google Cloud project and region:
PROJECT_ID = "gemini-api-473307"  # Replace with your actual project ID
REGION = "global"                   # Or a specific supported region (e.g., 'europe-west1')

# Authenticate via `gcloud auth application-default login` or set GOOGLE_APPLICATION_CREDENTIALS
client = AnthropicVertex(project_id=PROJECT_ID, region=REGION)

response = client.messages.create(
    model="claude-sonnet-4@20250514",  # The current Claude Sonnet 4 model id for Vertex AI
    max_tokens=1024,
    messages=[
        {"role": "user", "content": "Give me a Python example for Fibonacci numbers."}
    ],
)

print(response.model_dump_json(indent=2))
```
